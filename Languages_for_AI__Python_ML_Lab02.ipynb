### Датасет Б: «Стоимость квартиры»

Синтетический датасет, имитирующий объявления о продаже квартир.
Целевая переменная — `price` (цена в условных единицах).

Признаки:
- `area` — площадь, м²;
- `rooms` — число комнат;
- `floor` — этаж;
- `total_floors` — этажность дома;
- `distance_to_metro` — расстояние до метро, мин;
- `year_built` — год постройки;
- `is_center` — бинарный признак «центр»;
- `has_balcony` — бинарный признак «есть балкон» (шумовой);
- `renovation_age` — сколько лет назад был ремонт (шумовой).

Запустите следующую ячейку, чтобы получить данные.

Полученные данные запишите в файл csv для дальнейшей работы (использовать pandas)

```python
import numpy as np
import pandas as pd

def make_apartments_data(n_samples=1500, random_state=42):
    rng = np.random.default_rng(random_state)

    area = rng.normal(60, 22, n_samples).clip(18, 220)
    rooms = rng.integers(1, 6, n_samples)
    floor = rng.integers(1, 26, n_samples)
    total_floors = floor + rng.integers(0, 10, n_samples)
    distance_to_metro = rng.exponential(7, n_samples).clip(0, 35)
    year_built = rng.integers(1950, 2024, n_samples)
    is_center = rng.binomial(1, 0.2, n_samples)

    # Шумовые признаки
    has_balcony = rng.binomial(1, 0.6, n_samples)
    renovation_age = rng.integers(0, 30, n_samples)

    noise = rng.normal(0, 500, n_samples)

    price = (
        80 * area
        + 400 * rooms
        + 15 * floor
        - 70 * distance_to_metro
        + 4 * (year_built - 1950)
        + 2500 * is_center
        + 300 * has_balcony
        - 20 * renovation_age
        + noise
    )

    X = pd.DataFrame({
        "area": area.round(1),
        "rooms": rooms,
        "floor": floor,
        "total_floors": total_floors,
        "distance_to_metro": distance_to_metro.round(1),
        "year_built": year_built,
        "is_center": is_center,
        "has_balcony": has_balcony,
        "renovation_age": renovation_age,
    })
    return X, pd.Series(price, name="price")

X_reg, y_reg = make_apartments_data()
X_reg.head()
```

### Датасет В: «Кредитный скоринг» 

Синтетический датасет о клиентах банка.
Целевая переменная — `default` (1 — клиент допустил дефолт, 0 — вернул кредит).

Признаки:
- `age` — возраст;
- `income` — годовой доход;
- `loan_amount` — сумма кредита;
- `loan_term_months` — срок кредита, мес.;
- `num_prev_loans` — число предыдущих кредитов;
- `num_late_payments` — число просрочек;
- `employment_years` — стаж на последнем месте работы.

Запустите следующую ячейку, чтобы получить данные. Полученные данные запишите в файл csv для дальнейшей работы (использовать pandas)


```python
def make_credit_data(n_samples=2000, random_state=42):
    rng = np.random.default_rng(random_state)

    age = rng.integers(20, 70, n_samples)
    income = rng.lognormal(mean=10.5, sigma=0.5, size=n_samples).round(0)
    loan_amount = rng.lognormal(mean=10.0, sigma=0.6, size=n_samples).round(0)
    loan_term_months = rng.choice([12, 24, 36, 48, 60], size=n_samples)
    num_prev_loans = rng.integers(0, 6, n_samples)
    num_late_payments = rng.integers(0, 8, n_samples)
    employment_years = rng.integers(0, 30, n_samples)

    logit = (
        -1.5
        + 0.5 * (loan_amount / income)
        - 0.05 * employment_years
        + 0.4 * num_late_payments
        + 0.15 * num_prev_loans
        - 0.02 * age
        + rng.normal(0, 0.5, n_samples)
    )
    prob_default = 1 / (1 + np.exp(-logit))
    default = rng.binomial(1, prob_default)

    X = pd.DataFrame({
        "age": age,
        "income": income,
        "loan_amount": loan_amount,
        "loan_term_months": loan_term_months,
        "num_prev_loans": num_prev_loans,
        "num_late_payments": num_late_payments,
        "employment_years": employment_years,
    })
    return X, pd.Series(default, name="default")

X_cls, y_cls = make_credit_data()
X_cls.head()
```
